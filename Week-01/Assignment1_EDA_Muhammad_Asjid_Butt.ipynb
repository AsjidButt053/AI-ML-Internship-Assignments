```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset

```python
df=pd.read_csv('Assignment_Dataset.csv.csv')
df.head()
```

## Inspect Dataset

```python
print(df.shape)
df.info()
df.describe(include='all')
```

### Initial Observations
- Dataset contains Airbnb listings.
- Mix of numeric and categorical columns.
- Missing values exist and require cleaning.

## Data Cleaning

```python
df=df.drop_duplicates()
if 'last_review' in df.columns:
    df['last_review']=pd.to_datetime(df['last_review'])
if 'reviews_per_month' in df.columns:
    df['reviews_per_month']=df['reviews_per_month'].fillna(df['reviews_per_month'].median())
if 'name' in df.columns:
    df['name']=df['name'].fillna('Unknown')
if 'host_name' in df.columns:
    df['host_name']=df['host_name'].fillna('Unknown')
df=df[df['price']>0]
df=df[df['minimum_nights']<365]
```

## Feature Engineering

```python
bins=[0,100,250,100000]
labels=['Budget','Mid-range','Luxury']
df['Price_Category']=pd.cut(df['price'],bins=bins,labels=labels)
```

## Univariate Analysis

```python
plt.figure(figsize=(6,4))
sns.histplot(df['price'],bins=50); plt.xlim(0,600); plt.show()
```

**Interpretation:** Price distribution shows right skew.

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df,x='room_type'); plt.xticks(rotation=20); plt.show()
```

**Interpretation:** Room type frequency.

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df,x='neighbourhood_group'); plt.show()
```

**Interpretation:** Listings by borough.

```python
plt.figure(figsize=(6,4))
sns.histplot(df['minimum_nights'],bins=30); plt.xlim(0,100); plt.show()
```

**Interpretation:** Minimum nights distribution.

## Bivariate & Multivariate Analysis

```python
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x='neighbourhood_group',y='price'); plt.ylim(0,600); plt.show()
```

**Interpretation:** Price by borough.

```python
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x='room_type',y='price'); plt.ylim(0,600); plt.show()
```

**Interpretation:** Price by room type.

```python
plt.figure(figsize=(7,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True,cmap='coolwarm'); plt.show()
```

**Interpretation:** Correlation heatmap.

```python
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x='neighbourhood_group',y='price',hue='room_type'); plt.ylim(0,600); plt.show()
```

**Interpretation:** Three-variable relationship.

## Final Insights
1. Price distribution is highly skewed.
2. Entire homes are generally more expensive.
3. Manhattan tends to have higher prices.
4. Availability and reviews have weak correlation with price.
5. **Recommendation:** New hosts should price competitively based on room type and borough.
