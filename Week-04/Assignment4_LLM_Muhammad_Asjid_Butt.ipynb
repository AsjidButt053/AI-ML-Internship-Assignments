# Lab 16 — Introduction to LLMs

## 1. What is an LLM?

A Large Language Model (LLM) is a neural network based on the Transformer architecture, trained on large amounts of text to predict the next token.

LLMs can perform tasks such as:
- Text generation
- Summarization
- Translation
- Question answering
- Conversation

## 2. Tokenization

LLMs process tokens rather than raw words. Tokens may be complete words, subwords, punctuation, or other pieces of text.

````python
!pip -q install transformers sentencepiece
````

````python
from transformers import AutoTokenizer

print("Transformers imported successfully!")
````

````python
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

text = "Unbelievable language models are powerful."

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print("Original text:")
print(text)

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)
````

````python
decoded = tokenizer.decode(token_ids)

print("Decoded text:")
print(decoded)
````

### Observation

The tokenizer converts human-readable text into smaller units and maps those units to numerical IDs, which can then be processed by the neural network.

## 3. Transformer and Self-Attention

Transformers use self-attention to determine which parts of an input are relevant to each other.

### Simplified workflow

```text

```

## 4. How LLMs are Trained

### Pretraining
The model learns general language patterns from a very large corpus by predicting the next token.

### Fine-tuning
A pretrained model is further trained on a smaller, more specific dataset or task.

### RLHF
Human feedback can be used to encourage preferred response behavior.

## 5. Context Window

A context window is the maximum amount of token information a model can consider at one time. Information outside the available context cannot directly be considered by the model.

## 7. LLM Limitations

1. **Hallucination** — the model can produce confident but incorrect information.
2. **Knowledge cutoff** — information beyond the training data may not be known.
3. **Bias** — training data can contain biases.
4. **Cost and latency** — larger models can require more computation.

# Lab 17 — Prompt Engineering

## 1. What is Prompt Engineering?

Prompt engineering is the practice of carefully designing the input given to an LLM so that it reliably produces the desired output.

Useful prompt elements include:
- Role
- Task
- Examples
- Output format
- Constraints
- Context

## 2. Zero-Shot Prompting

Zero-shot prompting asks the model to perform a task without providing examples.

**Prompt:**

> Classify this review as positive or negative: "The battery life is terrible."

**Solved output:** Negative

````python
zero_shot_prompt = '''
Classify the following review as Positive or Negative.

Review: "The battery life is terrible."

Return only one word.
'''

print(zero_shot_prompt)
print("\nSolved output: Negative")
````

## 3. Few-Shot Prompting

Few-shot prompting provides example input-output pairs before the actual question.

````python
few_shot_prompt = '''
Review: "I loved this product!" -> Positive
Review: "Completely disappointed." -> Negative
Review: "The battery life is terrible." ->
'''

print(few_shot_prompt)
print("\nSolved output: Negative")
````

## 4. Step-by-Step Reasoning Prompt

Example problem:

A store had 120 apples. They sold 35% in the morning and 28 in the afternoon.

````python
starting_apples = 120

morning_sales = starting_apples * 0.35
after_morning = starting_apples - morning_sales

afternoon_sales = 28
remaining = after_morning - afternoon_sales

print("Morning sales:", int(morning_sales))
print("After morning:", int(after_morning))
print("Afternoon sales:", afternoon_sales)
print("Apples remaining:", int(remaining))
````

### Solved Result

- Morning sales = 35% × 120 = 42
- Remaining after morning = 120 − 42 = 78
- Afternoon sales = 28
- Final answer = **50 apples**

## 7. Temperature

- **Low temperature (around 0.1–0.3):** generally more focused and repeatable.
- **Medium temperature (around 0.5):** balanced behavior.
- **High temperature (around 0.8–1.0):** generally more varied and creative.

The appropriate value depends on the task.

````python
temperatures = [0.2, 0.5, 0.9]

for temperature in temperatures:
    if temperature <= 0.3:
        behavior = "focused / deterministic"
    elif temperature >= 0.8:
        behavior = "creative / varied"
    else:
        behavior = "balanced"

    print(f"Temperature = {temperature}: {behavior}")
````

## 8. Weak Prompt vs Strong Prompt

### Weak Prompt

```text
Summarize this meeting.
```

### Strong Prompt

```text
You are a professional executive assistant.
Summarize the meeting transcript in exactly 3 bullet points.
Focus only on decisions and action items.
For each action item, include the responsible person if mentioned.
Ignore small talk.
```

The strong prompt specifies role, task, output length, focus, and constraints.

````python
weak_prompt = "Summarize this meeting."

strong_prompt = '''
You are a professional executive assistant.
Summarize the meeting transcript in exactly 3 bullet points.
Focus only on decisions and action items.
For each action item, include the responsible person if mentioned.
Ignore small talk.
'''

print("WEAK PROMPT:")
print(weak_prompt)

print("\nSTRONG PROMPT:")
print(strong_prompt)
````
